**1. Loading the Dataset & Imports**

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Setting visual style for the plots
sns.set_theme(style="whitegrid")

# Loading the dataset
df = pd.read_csv('horse.csv')
```

**2. Basic Statistical Analysis**

```python
# Display first and last 6 records
display(df.head(6))
display(df.tail(6))

# Dataset info (data types, non-null counts)
df.info()

# Basic statistical summary for numerical columns
display(df.describe())

# Basic statistical summary for categorical columns
display(df.describe(include=['object', 'category']))
```

**3. Handling Missing Data**

```python
# Identifying total missing values per column
missing_counts = df.isnull().sum()
print("Missing Values:\n", missing_counts[missing_counts > 0])

# Display records containing At Least one missing value
missing_rows = df[df.isnull().any(axis=1)]
display(missing_rows)

# Remove columns with more than 70% missing values
threshold = 0.7 * len(df)
df = df.dropna(thresh=threshold, axis=1)

# Replace missing numerical values with the mean
numeric_cols = df.select_dtypes(include=[np.number]).columns
df[numeric_cols] = df[numeric_cols].fillna(df[numeric_cols].mean())

# For categorical columns, you can fill with the mode
cat_cols = df.select_dtypes(include=['object', 'category']).columns
for col in cat_cols:
    df[col] = df[col].fillna(df[col].mode()[0])
```

**4. Data Cleaning**

```python
# Identify and remove duplicate rows
print(f"Duplicate rows found: {df.duplicated().sum()}")
df = df.drop_duplicates()

# Get a list of all columns that contain text (object data type)
text_columns = df.select_dtypes(include=['object']).columns

# Loop through each text column to clean and convert it
for col in text_columns:
    # Convert all text in the column to lowercase to ensure consistency
    # (this fixes any issues where some words are capitalized and others are not)
    df[col] = df[col].str.lower()

    # Convert the cleaned character variable into a categorical type
    df[col] = df[col].astype('category')

# Display the updated data types just to verify my changes worked
print("\nUpdated Data Types:")
print(df.dtypes)
```

**5. Data Transformation**

```python
# Import tool to convert text categories into numbers
from sklearn.preprocessing import LabelEncoder
le = LabelEncoder()

# Encode 'outcome' and 'surgery' so I can use them for correlation analysis later
df['outcome_encoded'] = le.fit_transform(df['outcome'])
df['surgery_encoded'] = le.fit_transform(df['surgery'])

# Draw a boxplot to check for outliers (using 'pulse' as my example)
plt.figure(figsize=(8, 4))
sns.boxplot(x=df['pulse'])
plt.title('Outlier Detection: Pulse Boxplot')
plt.show()

# Function to calculate and remove extreme outliers using the IQR formula
def remove_outliers_iqr(dataframe, column):
    Q1 = dataframe[column].quantile(0.25)
    Q3 = dataframe[column].quantile(0.75)
    IQR = Q3 - Q1

    # Calculate my upper and lower limits
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    # Keeping only the rows that falls inside these limits
    return dataframe[(dataframe[column] >= lower_bound) & (dataframe[column] <= upper_bound)]

# Applying my function to drop outliers from the pulse column
df = remove_outliers_iqr(df, 'pulse')

print("Data transformation complete. Outliers removed and new encoded columns added.")
display(df.head())
```

**6. Univariate Analysis (Min 3 Visualizations)**

```python
# 1. Histogram with Density Curve (KDE) - Numeric
# I am using the 'pulse' column to see the distribution of the horse's heart rates
plt.figure(figsize=(8, 5))
sns.histplot(df['pulse'], kde=True, bins=20, color='blue')
plt.title('Distribution of Pulse Rates')
plt.xlabel('Pulse')
plt.ylabel('Count')
plt.show()

# 2. Percentage Bar Chart - Categorical
# Percentage breakdown of the 'outcome' categories
plt.figure(figsize=(8, 5))
cat_counts = df['outcome'].value_counts(normalize=True) * 100
sns.barplot(x=cat_counts.index, y=cat_counts.values, palette='viridis')
plt.title('Percentage Distribution of Outcomes')
plt.ylabel('Percentage (%)')
plt.xlabel('Outcome')
plt.show()

# 3. Violin Plot - Numeric
# Using 'rectal_temp' to check the spread and density of body temperatures
plt.figure(figsize=(8, 5))
sns.violinplot(x=df['rectal_temp'], color='lightgreen')
plt.title('Violin Plot of Rectal Temperature')
plt.xlabel('Rectal Temperature')
plt.show()
```

**7. Bivariate Analysis (Min 3 Visualizations)**

```python
# 1. Scatter Plot (Numeric vs Numeric)
# I am comparing rectal temperature and pulse to see if there is any relationship between them
plt.figure(figsize=(8, 5))
sns.scatterplot(data=df, x='rectal_temp', y='pulse', alpha=0.7)
plt.title('Scatter Plot: Rectal Temp vs Pulse')
plt.xlabel('Rectal Temperature')
plt.ylabel('Pulse')
plt.show()

# 2. Boxplot (Categorical vs Numeric)
# I want to check how the pulse distribution looks based on the final outcome
plt.figure(figsize=(8, 5))
sns.boxplot(data=df, x='outcome', y='pulse', palette='Set2')
plt.title('Pulse Distribution Across Outcomes')
plt.xlabel('Outcome')
plt.ylabel('Pulse')
plt.show()

# 3. Bar Chart (Categorical vs Numeric - Mean)
# I am calculating the average pulse for horses that did and didn't have surgery
plt.figure(figsize=(8, 5))
sns.barplot(data=df, x='surgery', y='pulse', estimator=np.mean, ci=None, palette='pastel')
plt.title('Average Pulse by Surgery Status')
plt.xlabel('Surgery')
plt.ylabel('Average Pulse')
plt.show()
```

**8. Multivariate Analysis (Min 3 Visualizations)**

```python
# 1. Heatmap (Correlation Matrix for all numeric columns)
# I am creating a heatmap to see how all the numeric variables correlate with each other
plt.figure(figsize=(12, 8))
correlation_matrix = df.select_dtypes(include=[np.number]).corr()
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', fmt=".2f")
plt.title('Correlation Heatmap')
plt.show()

# 2. FacetGrid / catplot
# I want to look at the distribution of pulse rates separately for each outcome group
g = sns.FacetGrid(df, col="outcome", height=4, aspect=1)
g.map(sns.histplot, "pulse", kde=True, color="purple")
g.fig.suptitle('FacetGrid: Pulse Distribution by Outcome', y=1.05)
plt.show()

# 3. Scatter Plot with Hue and Size (4 variables)
# Let's compare rectal temp and pulse, coloring by outcome and sizing the points by packed cell volume
plt.figure(figsize=(9, 6))
sns.scatterplot(data=df,
                x='rectal_temp',
                y='pulse',
                hue='outcome',
                size='packed_cell_volume',
                sizes=(20, 200),
                alpha=0.7,
                palette='Set1')
plt.title('Rectal Temp vs Pulse (Colored by Outcome, Sized by PCV)')
plt.xlabel('Rectal Temperature')
plt.ylabel('Pulse')
plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left')
plt.show()
```
