### **PHASE 1: DATA PREPROCESSING & INITIAL EDA**

**1. Loading the Dataset & Imports**

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Setting visual style for the plots
sns.set_theme(style="whitegrid")

# Loading the dataset directly from GitHub (using the Raw URL)
url = "https://raw.githubusercontent.com/Rudragupta8777/Rudra-Gupta/main/horse.csv"
df = pd.read_csv(url)
```

**2. Basic Statistical Analysis**

```python
# Display first and last 6 records
display(df.head(6))
display(df.tail(6))

# Dataset info (data types, non-null counts)
df.info()

# Basic statistical summary for numerical columns
display(df.describe())

# Basic statistical summary for categorical columns
display(df.describe(include=['object', 'category']))
```

**3. Handling Missing Data**

```python
# Identifying total missing values per column
missing_counts = df.isnull().sum()
print("Missing Values:\n", missing_counts[missing_counts > 0])

# Display records containing At Least one missing value
missing_rows = df[df.isnull().any(axis=1)]
display(missing_rows)

# Remove columns with more than 70% missing values
threshold = 0.7 * len(df)
df = df.dropna(thresh=threshold, axis=1)

# Replace missing numerical values with the mean
numeric_cols = df.select_dtypes(include=[np.number]).columns
df[numeric_cols] = df[numeric_cols].fillna(df[numeric_cols].mean())

# For categorical columns, you can fill with the mode
cat_cols = df.select_dtypes(include=['object', 'category']).columns
for col in cat_cols:
    df[col] = df[col].fillna(df[col].mode()[0])
```

**4. Data Cleaning**

```python
# Identify and remove duplicate rows
print(f"Duplicate rows found: {df.duplicated().sum()}")
df = df.drop_duplicates()

# Get a list of all columns that contain text (object data type)
text_columns = df.select_dtypes(include=['object']).columns

# Loop through each text column to clean and convert it
for col in text_columns:
    # Convert all text in the column to lowercase to ensure consistency
    # (this fixes any issues where some words are capitalized and others are not)
    df[col] = df[col].str.lower()

    # Convert the cleaned character variable into a categorical type
    df[col] = df[col].astype('category')

# Display the updated data types just to verify my changes worked
print("\nUpdated Data Types:")
print(df.dtypes)
```

**5. Data Transformation**

```python
# Import tool to convert text categories into numbers
from sklearn.preprocessing import LabelEncoder
le = LabelEncoder()

# Encode 'outcome' and 'surgery' so I can use them for correlation analysis later
df['outcome_encoded'] = le.fit_transform(df['outcome'])
df['surgery_encoded'] = le.fit_transform(df['surgery'])

# Draw a boxplot to check for outliers (using 'pulse' as my example)
plt.figure(figsize=(8, 4))
sns.boxplot(x=df['pulse'])
plt.title('Outlier Detection: Pulse Boxplot')
plt.show()

# Function to calculate and remove extreme outliers using the IQR formula
def remove_outliers_iqr(dataframe, column):
    Q1 = dataframe[column].quantile(0.25)
    Q3 = dataframe[column].quantile(0.75)
    IQR = Q3 - Q1

    # Calculate my upper and lower limits
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    # Keeping only the rows that falls inside these limits
    return dataframe[(dataframe[column] >= lower_bound) & (dataframe[column] <= upper_bound)]

# Applying my function to drop outliers from the pulse column
df = remove_outliers_iqr(df, 'pulse')

print("Data transformation complete. Outliers removed and new encoded columns added.")
display(df.head())
```

**6. Univariate Analysis (Min 3 Visualizations)**

```python
# 1. Histogram with Density Curve (KDE) - Numeric
# I am using the 'pulse' column to see the distribution of the horse's heart rates
plt.figure(figsize=(8, 5))
sns.histplot(df['pulse'], kde=True, bins=20, color='blue')
plt.title('Distribution of Pulse Rates')
plt.xlabel('Pulse')
plt.ylabel('Count')
plt.show()

# 2. Percentage Bar Chart - Categorical
# Percentage breakdown of the 'outcome' categories
plt.figure(figsize=(8, 5))
cat_counts = df['outcome'].value_counts(normalize=True) * 100
sns.barplot(x=cat_counts.index, y=cat_counts.values, palette='viridis')
plt.title('Percentage Distribution of Outcomes')
plt.ylabel('Percentage (%)')
plt.xlabel('Outcome')
plt.show()

# 3. Violin Plot - Numeric
# Using 'rectal_temp' to check the spread and density of body temperatures
plt.figure(figsize=(8, 5))
sns.violinplot(x=df['rectal_temp'], color='lightgreen')
plt.title('Violin Plot of Rectal Temperature')
plt.xlabel('Rectal Temperature')
plt.show()
```

**7. Bivariate Analysis (Min 3 Visualizations)**

```python
# 1. Scatter Plot (Numeric vs Numeric)
# I am comparing rectal temperature and pulse to see if there is any relationship between them
plt.figure(figsize=(8, 5))
sns.scatterplot(data=df, x='rectal_temp', y='pulse', alpha=0.7)
plt.title('Scatter Plot: Rectal Temp vs Pulse')
plt.xlabel('Rectal Temperature')
plt.ylabel('Pulse')
plt.show()

# 2. Boxplot (Categorical vs Numeric)
# I want to check how the pulse distribution looks based on the final outcome
plt.figure(figsize=(8, 5))
sns.boxplot(data=df, x='outcome', y='pulse', palette='Set2')
plt.title('Pulse Distribution Across Outcomes')
plt.xlabel('Outcome')
plt.ylabel('Pulse')
plt.show()

# 3. Bar Chart (Categorical vs Numeric - Mean)
# I am calculating the average pulse for horses that did and didn't have surgery
plt.figure(figsize=(8, 5))
sns.barplot(data=df, x='surgery', y='pulse', estimator=np.mean, ci=None, palette='pastel')
plt.title('Average Pulse by Surgery Status')
plt.xlabel('Surgery')
plt.ylabel('Average Pulse')
plt.show()
```

**8. Multivariate Analysis (Min 3 Visualizations)**

```python
# 1. Heatmap (Correlation Matrix for all numeric columns)
# I am creating a heatmap to see how all the numeric variables correlate with each other
plt.figure(figsize=(12, 8))
correlation_matrix = df.select_dtypes(include=[np.number]).corr()
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', fmt=".2f")
plt.title('Correlation Heatmap')
plt.show()

# 2. FacetGrid / catplot
# I want to look at the distribution of pulse rates separately for each outcome group
g = sns.FacetGrid(df, col="outcome", height=4, aspect=1)
g.map(sns.histplot, "pulse", kde=True, color="purple")
g.fig.suptitle('FacetGrid: Pulse Distribution by Outcome', y=1.05)
plt.show()

# 3. Scatter Plot with Hue and Size (4 variables)
# Let's compare rectal temp and pulse, coloring by outcome and sizing the points by packed cell volume
plt.figure(figsize=(9, 6))
sns.scatterplot(data=df,
                x='rectal_temp',
                y='pulse',
                hue='outcome',
                size='packed_cell_volume',
                sizes=(20, 200),
                alpha=0.7,
                palette='Set1')
plt.title('Rectal Temp vs Pulse (Colored by Outcome, Sized by PCV)')
plt.xlabel('Rectal Temperature')
plt.ylabel('Pulse')
plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left')
plt.show()
```

### **PHASE 2: STATISTICAL ANALYSIS & CLUSTERING**

```python
import scipy.stats as stats
import scipy.cluster.hierarchy as sch
from sklearn.cluster import KMeans, AgglomerativeClustering
from sklearn.preprocessing import StandardScaler
```

**9. 1D Statistical Analysis**

```python
# I am creating a structured dataframe to store all summary statistics for numeric variables
numeric_cols = ['rectal_temp', 'pulse', 'packed_cell_volume']
stats_df = pd.DataFrame({
    'Mean': df[numeric_cols].mean(),
    'Median': df[numeric_cols].median(),
    'Min': df[numeric_cols].min(),
    'Max': df[numeric_cols].max(),
    'Range': df[numeric_cols].max() - df[numeric_cols].min(),
    'Variance': df[numeric_cols].var(),
    'Std_Dev': df[numeric_cols].std(),
    'IQR': df[numeric_cols].apply(stats.iqr),
    'Skewness': df[numeric_cols].apply(stats.skew, nan_policy='omit'),
    'Kurtosis': df[numeric_cols].apply(stats.kurtosis, nan_policy='omit')
})
print("--- 1D Summary Statistics ---")
display(stats_df)

# Frequency and relative frequency for a categorical variable ('outcome')
freq_table = df['outcome'].value_counts()
rel_freq_table = df['outcome'].value_counts(normalize=True) * 100
print("\n--- Outcome Frequency Table ---\n", freq_table)

# Visualizing categorical distribution: Pie and Bar charts
fig, axes = plt.subplots(1, 2, figsize=(12, 5))
axes[0].pie(freq_table, labels=freq_table.index, autopct='%1.1f%%', colors=sns.color_palette("pastel"))
axes[0].set_title('Outcome Distribution (Pie Chart)')
sns.barplot(x=freq_table.index, y=freq_table.values, ax=axes[1], palette="viridis")
axes[1].set_title('Outcome Distribution (Bar Plot)')
plt.show()

# Frequency table with bin intervals for a numerical variable ('pulse')
df['pulse_bins'] = pd.cut(df['pulse'], bins=5)
print("\n--- Pulse Binned Frequency Table ---\n", df['pulse_bins'].value_counts().sort_index())

# Visualizing numerical distribution: Histogram and Boxplot
fig, axes = plt.subplots(1, 2, figsize=(12, 5))
sns.histplot(df['pulse'], bins=10, kde=True, ax=axes[0], color='blue')
axes[0].set_title('Pulse Histogram')
sns.boxplot(x=df['pulse'], ax=axes[1], color='lightblue')
axes[1].set_title('Pulse Box Plot')
plt.show()
```

**10. 2D Statistical Analysis**

```python
# Contingency table for two categorical variables with row/column percentages
cross_tab = pd.crosstab(df['surgery'], df['outcome'], margins=True)
row_perc = pd.crosstab(df['surgery'], df['outcome'], normalize='index') * 100
print("\n--- Contingency Table (Row Percentages) ---\n", row_perc)

# Stacked bar plot for the contingency table
row_perc.plot(kind='bar', stacked=True, figsize=(8, 5), colormap='Set2')
plt.title('Outcome Percentages by Surgery Status')
plt.ylabel('Percentage')
plt.show()

# Grouping numerical data by category
grouped_stats = df.groupby('outcome')['pulse'].agg(['mean', 'median', 'std'])
print("\n--- Pulse Stats Grouped by Outcome ---\n", grouped_stats)

# Side-by-side Box Plot and Violin Plot
fig, axes = plt.subplots(1, 2, figsize=(12, 5))
sns.boxplot(data=df, x='outcome', y='pulse', ax=axes[0], palette='pastel')
axes[0].set_title('Pulse across Outcomes (Box)')
sns.violinplot(data=df, x='outcome', y='pulse', ax=axes[1], palette='pastel')
axes[1].set_title('Pulse across Outcomes (Violin)')
plt.show()

# Pearson and Spearman correlation between two numeric variables
pearson_corr = df['rectal_temp'].corr(df['pulse'], method='pearson')
spearman_corr = df['rectal_temp'].corr(df['pulse'], method='spearman')
print(f"\nPearson Correlation (Temp & Pulse): {pearson_corr:.3f}")
print(f"Spearman Correlation (Temp & Pulse): {spearman_corr:.3f}")

# Scatter plot with regression line
plt.figure(figsize=(8, 5))
sns.regplot(data=df, x='rectal_temp', y='pulse', scatter_kws={'alpha':0.6}, line_kws={'color':'red'})
plt.title('Linear Regression: Rectal Temp vs Pulse')
plt.show()
```

**12. ND / Multivariate Analysis**

```python
# Stats across combinations of two categorical variables
multi_group_stats = df.groupby(['surgery', 'outcome'])['pulse'].agg(['mean', 'std'])
print("\n--- Pulse Stats across Surgery and Outcome ---\n", multi_group_stats)

# Faceted Box Plot and Grouped Bar Chart with Error Bars
sns.catplot(data=df, x='outcome', y='pulse', col='surgery', kind='box', palette='Set3', height=5, aspect=1)
plt.subplots_adjust(top=0.85)
plt.suptitle('Faceted Box Plot: Pulse by Outcome (Split by Surgery)')
plt.show()

plt.figure(figsize=(8, 5))
# I am using a grouped bar chart to show the mean pulse with 95% confidence intervals
sns.barplot(data=df, x='outcome', y='pulse', hue='surgery', palette='muted')
plt.title('Grouped Bar Chart with Error Bars (95% CI)')
plt.show()

# Correlation matrix computed separately for each category level
print("\n--- Correlation (Temp vs Pulse) for Survived vs Died ---")
for category in df['outcome'].dropna().unique():
    subset = df[df['outcome'] == category]
    corr_val = subset['rectal_temp'].corr(subset['pulse'])
    print(f"Outcome '{category}': {corr_val:.3f}")

# Full Correlation Matrix and Heatmap (using all numeric columns)
plt.figure(figsize=(8, 6))
full_corr = df[numeric_cols].corr()
sns.heatmap(full_corr, annot=True, cmap='coolwarm', fmt=".2f", vmin=-1, vmax=1)
plt.title('Full Correlation Heatmap')
plt.show()

# Pairs Plot / Scatterplot Matrix to see all pairwise relationships at once
sns.pairplot(df[numeric_cols], diag_kind='kde', plot_kws={'alpha': 0.6})
plt.suptitle('Pairs Plot Matrix', y=1.02)
plt.show()
```

**13. Clustering Preparation**

```python
# Selecting only numerical features and handling any remaining missing values for clustering
cluster_data = df[['rectal_temp', 'pulse', 'packed_cell_volume']].dropna()

# Standardizing the data (scale)
scaler = StandardScaler()
scaled_features = scaler.fit_transform(cluster_data)
```

**14. K-Means Clustering**

```python
# Applying K-Means for center as 2
kmeans_2 = KMeans(n_clusters=2, random_state=42, n_init=10)
cluster_data['kmeans_2'] = kmeans_2.fit_predict(scaled_features)

print("\n--- K-Means (k=2) Cluster Centers (Scaled) ---")
print(kmeans_2.cluster_centers_)

# Calculating WCSS for K = 1 to 10 to find the Elbow
wcss = []
for i in range(1, 11):
    kmeans_temp = KMeans(n_clusters=i, random_state=42, n_init=10)
    kmeans_temp.fit(scaled_features)
    wcss.append(kmeans_temp.inertia_)

# Displaying the elbow plot
plt.figure(figsize=(8, 5))
plt.plot(range(1, 11), wcss, marker='o', linestyle='--', color='b')
plt.title('Elbow Method For Optimal k')
plt.xlabel('Number of Clusters (K)')
plt.ylabel('WCSS')
plt.xticks(range(1, 11))
plt.grid(True)
plt.show()

# Applying K-Means based on elbow recommendation (I will choose k=3 based on typical distributions here)
optimal_k = 3
kmeans_final = KMeans(n_clusters=optimal_k, random_state=42, n_init=10)
cluster_data['kmeans_optimal'] = kmeans_final.fit_predict(scaled_features)

plt.figure(figsize=(8, 5))
sns.scatterplot(data=cluster_data, x='rectal_temp', y='pulse', hue='kmeans_optimal', palette='Set1', alpha=0.8)
plt.title(f'K-Means Clustering with Optimal k={optimal_k}')
plt.show()
```

**15. Hierarchical Clustering**

```python
# Plotting dendrograms for Single, Complete, and Ward linkage methods
fig, axes = plt.subplots(1, 3, figsize=(18, 5))

linkages = ['single', 'complete', 'ward']
for i, method in enumerate(linkages):
    dendrogram = sch.dendrogram(sch.linkage(scaled_features, method=method), ax=axes[i], no_labels=True)
    axes[i].set_title(f'Dendrogram ({method.capitalize()} Linkage)')
    axes[i].set_xlabel('Samples')
    axes[i].set_ylabel('Distance')

plt.tight_layout()
plt.show()

# Dividing the data into 3 clusters using Agglomerative Clustering (Ward's method)
hc = AgglomerativeClustering(n_clusters=3, metric='euclidean', linkage='ward')
cluster_data['hierarchical_3'] = hc.fit_predict(scaled_features)

# Displaying the number of samples in each cluster (table equivalent)
print("\n--- Hierarchical Clustering Assignment Counts (k=3) ---")
print(cluster_data['hierarchical_3'].value_counts().sort_index())
```
